```python
import pandas as pd

train = pd.read_csv("/content/drive/MyDrive/FUTURE_ML_1/set3/train.csv")
test = pd.read_csv("/content/drive/MyDrive/FUTURE_ML_1/set3/test.csv")
store = pd.read_csv("/content/drive/MyDrive/FUTURE_ML_1/set3/store.csv")
```

```python
train_merged = train.merge(store, on='Store', how='left')
test_merged = test.merge(store, on='Store', how='left')
```

```python
train_merged['Date'] = pd.to_datetime(train_merged['Date'])
```

```python
train_merged.isnull().sum()
```

```python
train_merged['CompetitionDistance'].fillna(
    train_merged['CompetitionDistance'].median(), inplace=True)

train_merged.fillna(0, inplace=True)
```

```python
train_merged = train_merged[train_merged['Open'] == 1]
```

```python
monthly_sales = (
    train_merged
    .groupby(pd.Grouper(key='Date', freq='M'))['Sales']
    .sum()
    .reset_index()
)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(x=monthly_sales['Sales'])
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))
plt.plot(monthly_sales['ds'], monthly_sales['y'])
plt.xlabel("Date")
plt.ylabel("Sales")
plt.title("Monthly Sales Trend Over Time")
plt.show()
```

```python
monthly_sales['year'] = monthly_sales['ds'].dt.year

yearly_sales = monthly_sales.groupby('year')['y'].sum()

yearly_sales.plot(kind='bar', figsize=(10,5), title="Yearly Sales Comparison")
plt.ylabel("Sales")
plt.show()
```

```python
monthly_sales['month'] = monthly_sales['ds'].dt.month

monthly_pattern = monthly_sales.groupby('month')['y'].mean()

monthly_pattern.plot(kind='bar', figsize=(10,5), title="Average Monthly Sales")
plt.ylabel("Sales")
plt.show()
```

```python
from prophet import Prophet
```

```python
model = Prophet(
    yearly_seasonality=True,
    weekly_seasonality=False,
    daily_seasonality=False
)
```

```python
model.fit(monthly_sales[['ds', 'y']])
```

```python
future = model.make_future_dataframe(periods=12, freq='M')
```

```python
forecast = model.predict(future)
```

```python
forecast[['ds', 'yhat', 'yhat_lower', 'yhat_upper']].tail()
```

```python
model.plot(forecast)
plt.title("Sales Forecast with Prophet")
plt.show()
```

```python
model.plot_components(forecast)
plt.show()
```
